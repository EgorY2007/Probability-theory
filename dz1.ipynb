# Тема 4. Условная вероятность

> Смоделируйте задачу. В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

* Рассмотрим событие $A$ - шар, вынутый из второй урны - белый.

```python
import random

urna1 = 'Б' * 15 + 'Ч' * 5 
urna2 = 'Б' * 6 + 'Ч'
cnt = 1000000
bel_cnt = 0
for i in range(cnt):
    balls = random.sample(urna1,k=3)
    urna3 = urna2 + ''.join(balls)
    schar = random.choice(urna3)
    if schar == 'Б':
        bel_cnt += 1
print(round(bel_cnt/cnt, 4))
```

# Тема 5. Повторные испытания в схеме Бернулли

## Задание 2.9

>  В среднем 20% пакетов акций на аукционах продаются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 1) не будут проданы 5 пакетов; 2) будет продано: а) менее 2 пакетов; б) не более 2 пакетов; в) хотя бы 2 пакета; г) наивероятнейшее число пакетов.

* Рассмотрим событие $A$ - не будут проданы 5 пакетов по первоначально заявленной цене

```python
n = 9
p = 0.8
m = 5

prob = binom.pmf(m, n, p)
print(f'Вероятность по Бернулли = {prob:.4f}')
```

* Рассмотрим событие $B$ - будет продано менее 2 пакетов по первоначально заявленной цене

```python
n = 9
p = 0.2
m = 2

prob = binom.cdf(m - 1, n, p)
print(f'Вероятность по Бернулли = {prob:.4f}')
```

* Рассмотрим событие $С$ - будет продано не более 2 пакетов по первоначально заявленной цене

```python
n = 9
p = 0.2
m = 2

prob = binom.cdf(m, n, p)
print(f'Вероятность по Бернулли = {prob:.4f}')
```

* Рассмотрим событие $D$ - будет продано хотя бы 2 пакета по первоначально заявленной цене

```python
n = 9
p = 0.2
m = 2

prob = 1 - binom.cdf(m - 1, n, p)
print(f'Вероятность по Бернулли = {prob:.4f}')
```

* Рассмотрим событие $E$ - будет продано наивероятнейшее число пакетов по первоначально заявленной цене

```python
n = 9
p = 0.2
q = 1 - p

prob = 0
k_min = int(np.ceil(n * p - q))
k_max = int(np.floor(n * p + p))
for k in range(k_min, k_max + 1):
    prob += binom.pmf(k, n, p)
print(f'Вероятность по Бернулли = {prob:.4f}')
```

## Задание 2.10

> Завод отправил на базу 10 000 стандартных изделий. Среднее число изделий, повреждаемых при транспортировке, составляет 0,02%. Найти вероятность того, что из 10 000 изделий:
> 1. будет повреждено: а) 3; б) по крайней мере 3;
> 2. не будет повреждено: а) 9997; б) хотя бы 9997.

* Рассмотрим событие $A$ - будет повреждено 3 изделия

```python
n = 10000
p = 0.0002
lmd = n * p
m = 3

prob = poisson.pmf(k=m, mu=lmd)
print(f'Вероятность по Пуассону = {prob:.4f}')
```

* Рассмотрим событие $B$ -  будет повреждено по крайней мере 3 изделия

```python
prob = 1 - poisson.cdf(m-1,lmd)
print(f'Вероятность по Пуассону = {prob:.4f}')
```

* Рассмотрим событие $C$ - не будет повреждено 9997 изделий
> P(C) = P(A)

```python
prob = poisson.pmf(k=m, mu=lmd)
print(f'Вероятность по Пуассону = {prob:.4f}')
```

* Рассмотрим событие $D$ - не будет повреждено хотя бы 9997 изделий

```python
prob = poisson.cdf(m,lmd)
print(f'Вероятность по Пуассону = {prob:.4f}')
```

# Задание 2.11

> Пример 2.11. По результатам проверок налоговыми инспекциями установлено, что в среднем каждое второе малое предприятие региона имеет нарушение финансовой дисциплины. Найти вероятность того, что из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины:
а) 480 предприятий;
б) наивероятнейшее число предприятий;
в) не менее 480;
г) от 480 до 520.

* Рассмотрим событие $A$ - из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины 480 предприятий

```python
n = 1000
p = 0.5
q = 1 - p
m = 480

x = (m - n * p) / np.sqrt(n*p*q)

prob = norm.pdf(x) / np.sqrt(n*p*q)
print(f'Вероятность по Муавра-Лапласа = {prob:.4f}')
```

* Рассмотрим событие $B$ - из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины наивероятнейшее число предприятий

```python
prob = 0
k_min = int(np.ceil(n * p - q))
k_max = int(np.floor(n * p + p))

for k in range(k_min, k_max + 1):
    x = (k - n * p) / np.sqrt(n*p*q)
    prob += norm.pdf(x) / np.sqrt(n*p*q)
    
print(f'Вероятность по Муавра-Лапласа = {prob:.4f}')
```

* Рассмотрим событие $C$ - из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины не менее 480

```python
a = 480
b = 1000

mu = n*p
sigma = np.sqrt(n*p*q)

x1 = (a - mu) / sigma
x2 = (b - mu) / sigma
prob = norm.cdf(x2) - norm.cdf(x1)
print(f'Вероятность по Муавра-Лапласа = {prob:.4f}')
```

* Рассмотрим событие $C$ - из 1000 зарегистрированных в регионе малых предприятий имеют нарушения финансовой дисциплины от 480 до 520

```python
a = 480
b = 520

mu = n*p
sigma = np.sqrt(n*p*q)

x1 = (a - mu) / sigma
x2 = (b - mu) / sigma
prob = norm.cdf(x2) - norm.cdf(x1)
print(f'Вероятность по Муавра-Лапласа = {prob:.4f}')
```
