* Рассмотрим событие $A$ - семизначного число из цифр 4, 5 и 6, в котором цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза

## Задача 1.10

* а) Рассмотрим событие $A$ - из карточек получилось слово "ТОР"

```python
round(1 / sc.perm(6,3), 4)
```

```python
round(1 / sc.factorial(6), 4)
```

## Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что:
а) никакие два лица одного пола не сядут рядом;
б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

* а) Рассмотрим событие $A$ - рассадка 10 человек с условием, что никакие два лица одного пола не сядут рядом

```python
round(sc.factorial(5) * 2**5 / sc.factorial(10), 4)
```

## Задание 1.48
> В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности:
а) выпадения 11 очков;
б) выигрыша.

* а) Рассмотрим событие $A$ - получили при бросании 3 игральных костей сумму очков равную 11

```python
broski = list(product(range(1, 7), repeat=3))
total = len(broski)

сnt_11 = sum(x + y + z == 11 for x,y,z in broski)
cnt_11 / total
```

* б) Рассмотрим событие $B$ - получили при бросании 3 игральных костей сумму очков, превосходящую 10

```python
сnt_win = sum(x + y + z > 10 for x,y,z in broski)
cnt_win / total
```
